# Section 2 - Hypothesis 1

Testing the first hypothesis using Spearman Rank Correltaion, because we want to know if as the studytime goes up, the final grade increases.

```python
rho, p_value = spearmanr(
    df_cleaned["studytime"],
    df_cleaned["G3"]
)

print(f"Spearman correlation: {rho:.3f}")
print(f"P-value: {p_value:.4f}")
```

These numbers show that there is a statistical signifcance between weekly study time and a higher final grade but they also show that it's weak, showing that studying is benficial but it isn't a driving force in getting a definitive final grade improvement. Student Support Groups can help students improve their final grade (G3) by some amount through encouraging the ones that study less than 5 hours to study for longer sum durations.

# Section 3 - Hypothesis 2

Likewise, H2 will also be subject to a Spearman Rank Correlation test.

```python
rho, p_value = spearmanr(
    df_cleaned["failures"],
    df_cleaned["G3"]
)

print(f"Spearman correlation: {rho:.3f}")
print(f"P-value: {p_value:.4f}")
```

The rho of -0.361 and a p-value of 0.000 shows there is a significant and negative association between the previous class failures and a lower mathematics grade overall. This is evidence in support of H2 that previous failure does impact a students final grade, and that that impact increases as they continue to fail.

# Section 4 - Hypothesis 3

The near 0 correlation value indicates that there's no monotinoic relatinship between a student's rate of absence and their final grade. With a high p-value compared to the alpha of 0.05 also declaring that there is no statistical significance betweem the 2 features. Within the scope of this dataset H3 is unsupported.
